```python
from keras.datasets import mnist
from keras import models, layers
from keras.utils import to_categorical

#取得資料
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

#資料預處理
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1)).astype('float32') / 255

#One-hot編碼
train_labels = to_categorical(train_labels)
test_labels = to_categorical(test_labels)

#建立CNN模型架構
network = models.Sequential()
#第1層卷積：提取初步特徵
network.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)))
network.add(layers.MaxPooling2D((2, 2))) #池化層
#第2層卷積：提取更複雜的特徵
network.add(layers.Conv2D(64, (3, 3), activation='relu'))
network.add(layers.MaxPooling2D((2, 2)))
#第3層卷積
network.add(layers.Conv2D(64, (3, 3), activation='relu'))
#攤平層：將3D的特徵圖拉平成1D
network.add(layers.Flatten())
#全連接層：負責最後的分類判斷
network.add(layers.Dense(64, activation='relu'))
network.add(layers.Dense(10, activation='softmax')) #輸出0-9

#編譯模型
network.compile(optimizer='rmsprop',
                loss='categorical_crossentropy',
                metrics=['accuracy'])

#開始訓練
print("開始訓練...")
network.fit(train_images, train_labels, epochs=5, batch_size=64)

#評估準確度
test_loss, test_acc = network.evaluate(test_images, test_labels)
print(f'測試集準確度: {test_acc:.4f}')

#儲存模型
network.save('mnist_model.keras')
print("模型儲存成功：mnist_model.keras")
```
